# Notebook 01: Financial Data Acquisition & Exploration

This notebook demonstrates universe selection, data ingestion, adjusted price alignment, simple vs log return distributions, and corporate action adjustments.


```python
import sys
sys.path.insert(0, '..')
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from src.data import DataLoader, load_universe_data

# Load universe data from config
price_df, returns_df, bench_returns, raw_dict = load_universe_data('../config.yaml')
print(f'Loaded {price_df.shape[1]} universe assets across {price_df.shape[0]} trading days.')
price_df.head()
```

### 1. Cumulative Asset Performance
Let's visualize the normalized price trajectory ($P_t / P_0$) of our 20-stock universe against the SPY benchmark.


```python
normalized_prices = price_df / price_df.iloc[0]
plt.figure(figsize=(12, 6))
plt.plot(normalized_prices, alpha=0.6, linewidth=1)
plt.plot(bench_returns.add(1).cumprod(), color='black', linewidth=2.5, label='SPY Benchmark')
plt.title('Normalized Asset Price Growth (2015-2024)', fontweight='bold')
plt.xlabel('Date')
plt.ylabel('Normalized Price ($1.00 base)')
plt.legend(['Universe Equities (20)', 'SPY Benchmark'])
plt.grid(True, alpha=0.3)
plt.show()
```

### 2. Daily Return Distribution & Volatility


```python
ann_vol = returns_df.std() * np.sqrt(252) * 100
ann_ret = ((1 + returns_df.mean())**252 - 1) * 100
summary_df = pd.DataFrame({'Annualized Return (%)': ann_ret, 'Annualized Volatility (%)': ann_vol})
summary_df.sort_values(by='Annualized Return (%)', ascending=False)
```
