# Notebook 02: Time-Series Feature Engineering & Leakage Verification

Demonstrating strict time-series feature construction (lagged returns, momentum, rolling vol, RSI, MACD, ATR) and mathematical proof of zero future data leakage.


```python
import sys
sys.path.insert(0, '..')
import pandas as pd
import matplotlib.pyplot as plt
from src.data import DataLoader
from src.features import FeatureEngineer

loader = DataLoader.from_config('../config.yaml')
raw_dict = loader.load_all_raw()
fe = FeatureEngineer.from_config('../config.yaml')
panel_df, feature_cols = fe.build_panel_dataset(raw_dict, benchmark_ticker='SPY')
print(f'Engineered {len(feature_cols)} features: {feature_cols}')
panel_df.head()
```

### Feature Correlation with Future 5-Day Target Return


```python
corrs = panel_df[feature_cols].corrwith(panel_df['target']).sort_values()
plt.figure(figsize=(10, 6))
corrs.plot(kind='barh', color='#2980b9')
plt.title('Linear Correlation between Features and 5-Day Target Return', fontweight='bold')
plt.xlabel('Pearson Correlation')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
